```python
"""
Notebook Name: 03_Data_Analysis.ipynb
Description: Performs EDA, feature engineering (GDP per capita, growth rates),
             and exports the final analytical dataset for Power BI.
Author: Data Analyst Mentor & Mentee
"""

import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")

# 1. Load the Cleaned Master Dataset
processed_file_path = os.path.join(
    "..", "data", "processed", "master_economic_data.csv"
)

# Fallback path if running from root directory instead of notebooks folder
if not os.path.exists(processed_file_path):
  processed_file_path = os.path.join("data", "processed", "master_economic_data.csv")

df = pd.read_csv(processed_file_path)

print("Dataset loaded successfully!")
print(f"Dataset Shape: {df.shape}")
display(df.head())

# 2. Basic Statistical Summary
print("\n--- Summary Statistics ---")
display(df.describe())

# 3. Feature Engineering: Calculating GDP per Capita and Annual GDP Growth
print("\n--- Engineering Features ---")

# Sort values to ensure correct chronological calculations for time series
df = df.sort_values(by=["country_code", "year"]).reset_index(drop=True)

# Calculate GDP per Capita (GDP / Population)
# Note: Check exact column names generated during your pivot step (e.g., 'Gross Domestic Product (current US$)' or 'GDP (current US$)')
# Let's inspect column names dynamically to avoid mismatch errors
print("Available columns:", df.columns.tolist())

# Assuming standard World Bank indicator names mapped during cleaning:
# Let's write robust column mapping or safe access
gdp_col = [c for c in df.columns if "GDP" in c and "current" in c]
pop_col = [c for c in df.columns if "Population" in c]

if gdp_col and pop_col:
  gdp_col_name = gdp_col[0]
  pop_col_name = pop_col[0]

  df["gdp_per_capita"] = df[gdp_col_name] / df[pop_col_name]
  print("Successfully calculated: gdp_per_capita")

# Calculate Annual GDP Growth Rate (%)
if gdp_col:
  df["gdp_annual_growth_pct"] = (
      df.groupby("country_code")[gdp_col_name].pct_change() * 100
  )
  print("Successfully calculated: gdp_annual_growth_pct")

# 4. Export Final Analytical Dataset
historical_dir = os.path.join("..", "data", "historical")
if not os.path.exists(os.path.dirname(historical_dir)):
  historical_dir = os.path.join("data", "historical")

os.makedirs(historical_dir, exist_ok=True)

output_analytical_file = os.path.join(
    historical_dir, "analytical_master_dataset.csv"
)
df.to_csv(output_analytical_file, index=False)

print(
    f"\nFinal analytical dataset successfully exported to:"
    f" {output_analytical_file}"
)
```
