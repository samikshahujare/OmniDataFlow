```python
"""
Script Name: 01_API_Testing.py / Jupyter Notebook Snippet
Description: Tests connection to the World Bank REST API and retrieves live GDP data.
Author: Data Analyst Mentor & Mentee
"""

import sys
import requests


def test_world_bank_api():
  """Sends a GET request to the World Bank API to fetch GDP data for India

  and prints the status code and sample response payload.
  """
  # World Bank API endpoint for India's GDP (Indicator code: NY.GDP.MKTP.CD)
  # Format parameter forces the response to be returned as JSON
  url = "http://api.worldbank.org/v2/country/IND/indicator/NY.GDP.MKTP.CD?format=json"

  print(f"Connecting to World Bank API endpoint:\n{url}\n")

  try:
    # Send an HTTP GET request with a timeout of 10 seconds
    response = requests.get(url, timeout=10)

    # Check if the HTTP status code indicates success
    if response.status_code == 200:
      print("Success! HTTP Status Code: 200 OK\n")

      # Parse the JSON payload returned by the API
      data = response.json()

      # World Bank API returns a list where index 0 is metadata and index 1 is the actual data records
      print(f"Metadata summary: {data[0]}")
      print(f"\nTotal records returned in this page: {len(data[1])}\n")

      # Display the first 3 records to inspect the structure
      print("Displaying the first 3 records of raw JSON data:")
      for i, record in enumerate(data[1][:3]):
        print(
            f"Record {i+1} -> Year: {record.get('date')} | Value:"
            f" {record.get('value')}"
        )

    else:
      print(
          f"Failed to retrieve data. HTTP Status Code: {response.status_code}"
      )

  except requests.exceptions.RequestException as e:
    print(f"An error occurred during the API request: {e}", file=sys.stderr)


if __name__ == "__main__":
  test_world_bank_api()
```
